```python
import boto3

# Connect to S3
s3 = boto3.client('s3')

# Define bucket name
bucket_name = 'flight-tracking-analytics-project-2026'

# Create bucket
s3.create_bucket(
    Bucket=bucket_name,
    CreateBucketConfiguration={
        'LocationConstraint': 'ap-south-1'
    }
)

print("S3 Bucket created successfully!")
```

```python
# Define local CSV file path
file_path = "C:/Users/chira/OneDrive/Desktop/RegEx/Project October/Data Analytics Project/Flight/cleaned_flight_data.csv"
bucket_name = 'flight-tracking-analytics-project-2026'
# Define destination file name in S3
s3_file_name = "cleaned_flight_data.csv"

# Upload file to S3
s3.upload_file(
    file_path,
    bucket_name,
    s3_file_name
)

print("File uploaded successfully!")
```

```python
response = s3.list_objects_v2(
    Bucket=bucket_name
)

for obj in response.get('Contents', []):
    print("File:", obj['Key'])
    print("Size:", obj['Size'], "bytes")
    # list_objects_v2() retrieves the objects stored inside your S3 bucket. This verifies that your upload was successful.
    
```
